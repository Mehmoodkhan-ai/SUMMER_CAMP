### Class Variable vs Instance Variable
| Class Variable               | Instance Variable                |
| ---------------------------- | -------------------------------- |
| Class ke andar banta hai     | `self` ke sath banta hai         |
| Sab objects share karte hain | Har object ki alag copy hoti hai |
| `helpline`                   | `campus`                         |
| `obj_count`                  | `staff_count`                    |


```python
class saylani():
    helpline = '0900078601' # static
 
    obj_count = 0 # dynaminc
    
    def __init__(self, campus, staff_count, courses_count): 
        self.campus = campus
        self.staff_count = staff_count
        self.courses_count = courses_count
        saylani.obj_count +=1

    def detail(self, courses): # methods
        print(f"{self.campus} have {self.staff_count} and courrses are {self.courses_count}")
        # course = map(lambda x:x,courses)
        # print(*course)
    @staticmethod
    def facilities():
       print("medical also")
```

```python
za = saylani("ZA IT park", 10, 5) # object initialization
```

```python
za.detail(["AI", "DE", "WEB"])
```

TASK: Create a class of cars with atleast 3 attributes and two methods such that once method should be static. Also class should have 2 class variables and one should dynamic

```python
class Car():
    # Class Variables
    chasis_no = '1234567'
    engine_type = 'Electric'

    # dynamic variable
    mileage = 0
    
    # initializer
    def __init__(self, color, car_type, model_year):
        # class attributes
        self.color = color
        self.car_type = car_type
        self.model_year = model_year

    # instance method
    def details(self):
        print(f" Color: {self.color}, Type: {self.car_type}, Year: {self.model_year}")

    # static method
    @staticmethod
    def calculate_mileage(num):
        Car.mileage += num
        print(f"Your car runs {Car.mileage} km")
        
```

```python
swift = Car('silver', 'Hatchback', '2020')
```

### Comma kya karta hai?
- Comma multiple values ko print karta hai aur automatically type casting karta hai.
- Matlab Python khud:
- int → string
- float → string
- bool → string
- mein convert kar deta hai print karte waqt.
- Agar comma na use karo:
- print("Student Age:" + age)
- To error aayega 
- Kyunki:
- "Student Age:"  → String
- 20              → Integer
- String aur integer ko direct + se join nahi kar sakte.

```python
#Method :1 Concatenation with , [, will do by defualt type casting]
print("Student Name :", name)
print("Student Age:", age)
print("Student Weight:", weight)
print("Student is Present: ", is_present)
```

```python
## Method 2: Concatenation with + operator
print("Student Name : "+ name)
print("Student Age:"+ str(age))
print("Student Weight:"+ str(weight))
print("Student is Present: "+ str(is_present))
```

```python
print("The sum of 2 plus 2 is " + 4)
# Kyuki + yahan concatenation ke liye use ho raha hai.
# Dekho:
# "The sum of 2 plus 2 is "
# Ye string hai.
# Aur:
# 4
# Ye integer (int) hai.
# Python string aur integer ko direct join nahi kar sakta.
```
