- content = []
- Khali list bana di.
- Ab hum is list mein data store karenge.
- csv.reader()
- Ye har line ko comma se tod deta hai.
- append()
- content.append(line)
- Har line list mein store hoti jayegi.
- [
- ['USD','280'],
- ['EUR','310'],
- ['PKR','289']
- ]
- Ab data rows aur columns mein aa gaya.

```python
content = []
with open('./exchange_rate.csv', 'r') as f:
    csv_content = csv.reader(f)
    for line in csv_content:
        content.append(line)
```

```python
competitions = []
with open('./competitions.csv', 'r') as f:
    content = csv.reader(f)
    # print(content)
    for line in content:
       competitions.append(line)
```

```python
competitions
```

```python
with open(file='./competitions.csv', mode='a', newline='') as f:
    write_handler = csv.writer(f, delimiter=",")
    write_handler.writerow(['2012', 'Asia Cup', 'Pakistan'])
```

```python
import json
try:
    with open('./read.json','r') as f:
        json.load(f)
except FileNotFoundError as e:
    print(e)
finally:
    print('success')
```

# Summary
| Keyword                    | Matlab                                                              |
| -------------------------- | ------------------------------------------------------------------- |
| `try`                      | Yahan error aa sakta hai.                                           |
| `open()`                   | File kholna.                                                        |
| `'r'`                      | Sirf file read karni hai.                                           |
| `json.load()`              | JSON file ko Python dictionary/list mein convert karta hai.         |
| `except FileNotFoundError` | Agar file na mile to error handle karo.                             |
| `e`                        | Error message ko store karta hai.                                   |
| `print(e)`                 | Error screen par dikhata hai.                                       |
| `finally`                  | Ye block **hamesha execute hota hai**, chahe error aaye ya na aaye. |
