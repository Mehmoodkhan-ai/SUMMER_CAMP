```python
class cars():
    def __init__(self,color,model,make_year):#constructor
        self.color=color
        self.model=model
        self.make_year=make_year
    def start(self):
        """
        start object belongs to class car
        arg : No argumnts
        return : No return
        """
        print("car can start")
    def stop(self):
        """ 
        stop object belongs to class car
        arg : No argumnts
        return : No return
        """
        print("car can stop")
```

```python
civic= cars("black","honda","2026") # class initilization / object creation
```

```python
print(type(civic))
```

TASK:Create a class of humans that have atleast 2 attributes and methods

```python
class humans():
    def __init__(self,name,color):
        self.name=name
        self.color=color
    def walk (self):
        """
        THIS WALK FUNCTION BELONGS TO HUMANS CLASS
        ARG:NO ARGUMENTS
        RETURN:NO RETURN
        """
        print(f"{self.name} can walk")
    def speak(self):
        """
        THIS WALK FUNCTION BELONGS TO HUMANS CLASS
        ARG:NO ARGUMENTS
        RETURN:NO RETURN
        """
        print(f"{self.name} can speak")
```

```python
obj=humans("hamza","fair")
```

```python
obj.color
```

```python
obj.name
```

```python
obj.speak()
```

```python
obj.walk()
```
