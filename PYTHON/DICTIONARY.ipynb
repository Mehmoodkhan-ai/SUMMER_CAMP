# ***Changeable***
Dictionaries are changeable, meaning that we can change, add or remove items after the dictionary has been created.

# ***Duplicates Not Allowed***
Dictionaries cannot have two items with the same key:

```python
thisdict = {
  "brand": "Ford",
  "model": "Mustang",
  "year": 1964,
  "year": 2020
}
print(thisdict)
```

Python mein year: 2020 is liye show hoga kyun ke dictionary mein same key do baar likhne par last wali value purani value ko replace kar deti hai.

# ***Python Collections (Arrays)***
There are four collection data types in the Python programming language:

List is a collection which is ordered and changeable. Allows duplicate members.

Tuple is a collection which is ordered and unchangeable. Allows duplicate members.

Set is a collection which is unordered, unchangeable*, and unindexed. No duplicate members.

Dictionary is a collection which is ordered** and changeable. No duplicate members.

# ***Accessing Items***
You can access the items of a dictionary by referring to its key name, inside square brackets:

```python
thisdict = {
  "brand": "Ford",
  "model": "Mustang",
  "year": 1964
}
x = thisdict["model"]
```

```python
x = thisdict.get("model")
```

```python
x
```

```python
y = thisdict.keys()
```

```python
y
```

Add a new item to the original dictionary, and see that the keys list gets updated as well:

```python
car = {
"brand": "Ford",
"model": "Mustang",
"year": 1964
}

x = car.keys()
print(x)
```

Python mein dictionary ki kisi key ko access ya add karne ke liye square brackets [] use hote hain.

```python

car["color"] = "white"

print(x)
```

# ***Change Values***
Tum dictionary ki kisi specific value ko uski key ka naam use karke change kar sakte ho.

# ***Update Dictionary***
The update() method will update the dictionary with the items from the given argument.

```python
thisdict = {
  "brand": "Ford",
  "model": "Mustang",
  "year": 1964
}
thisdict.update({"year": 2020})
```

```python
thisdict
```

# ***Adding Items***
Adding an item to the dictionary is done by using a new index key and assigning a value to it:

```python
thisdict = {
  "brand": "Ford",
  "model": "Mustang",
  "year": 1964
}
thisdict["color"] = "red"
print(thisdict)
```

# ***Update Dictionary***
Python mein update() method dictionary ki existing values ko update karta hai. Agar koi key dictionary mein mojood nahi hai, to woh nayi key add kar deta hai.

```python
thisdict = {
  "brand": "Ford",
  "model": "Mustang",
  "year": 1964
}
thisdict.update({"color": "red"})
```

# ***Removing Items***
The pop() method removes the item with the specified key name:

```python
thisdict = {
  "brand": "Ford",
  "model": "Mustang",
  "year": 1964
}
thisdict.pop("model")
print(thisdict)
```

# ***POP ITEM***
Matlab: Python mein popitem() method dictionary mein se aakhri add kiya gaya key-value pair remove kar deta hai.

```python
thisdict = {
  "brand": "Ford",
  "model": "Mustang",
  "year": 1964
}
thisdict.popitem()
print(thisdict)
```

# ***DEL***
The del keyword removes the item with the specified key name:

```python
thisdict = {
  "brand": "Ford",
  "model": "Mustang",
  "year": 1964
}
del thisdict["model"]
print(thisdict)
```

# ***CLEAR()***
Python mein clear() method dictionary ke andar se saara data remove kar deta hai. Lekin dictionary khud mojood rehti hai.

```python
thisdict = {
  "brand": "Ford",
  "model": "Mustang",
  "year": 1964
}
thisdict.clear()
print(thisdict)
```

# ***Copy a Dictionary***
Make a copy of a dictionary with the copy() method:

```python
thisdict = {
  "brand": "Ford",
  "model": "Mustang",
  "year": 1964
}
mydict = thisdict.copy()
print(mydict)
```

```python
## Make a copy of a dictionary with the dict() function:

thisdict = {
  "brand": "Ford",
  "model": "Mustang",
  "year": 1964
}
mydict = dict(thisdict)
print(mydict)
```
