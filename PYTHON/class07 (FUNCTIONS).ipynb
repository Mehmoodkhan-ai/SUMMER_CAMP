### working with flexible number of arguments using `*`

```python
#Yahan 3 types ke parameters hain:

# num1 → Required
# num2 → Required
# *nums → Baqi jitni bhi values hongi, sab is mein aa jayengi tuple ki form mein.
def add(num1, num2, *nums):
    print(num1)
    print(num2)
    # Yahan * tuple ko unpack karta hai.
    print(*nums)
    added = num1+ num2
    
    for i in nums:
        added += i
    
    print(added)
```

num1 = 10
num2 = 20
nums = ()
Yani nums empty tuple hai.
nums hamesha tuple hota hai.

```python
add(3,4,1)
```

### working with flexible number of arguments using `**`

```python
def add(num1, num2, **nums):
    print(num1)
    print(num2)
    # yaha nums tuple nhi han balka dictionary han
    print(nums)
    added = num1 + num2
    # .items() dictionary ke key aur value dono deta hai.
    for key, value in nums.items():
        added += value 
    print(added)
```

```python
add(3,4)
```

```python
add(3,4, num3=5,num4=6)
```
